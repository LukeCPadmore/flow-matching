```python
import torchvision
from torchvision.transforms import v2
import torch

tfm = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
])

ds = torchvision.datasets.CIFAR10(
    root="data",
    train=True,
    download=True,
    transform=tfm,
)

x, y = ds[0]
print(x.shape, x.dtype, x.min(), x.max())
```

```python
import kornia

x = x.unsqueeze(0)  # [1, 3, 32, 32]
lab = kornia.color.rgb_to_lab(x)

print(lab.shape, lab.dtype, lab.min(), lab.max())
```
